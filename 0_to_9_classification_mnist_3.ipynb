```python
import numpy as np
import matplotlib.pyplot as plt 
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Layer,Dense,Input,Flatten,Conv2D
from tensorflow.keras.datasets.mnist import load_data
```

```python
def main():
    (trainx,trainy),(testx,testy)=load_data()
    model=model_b()
    history=model.fit(trainx,trainy,validation_split=0.1,epochs=1,batch_size=32,verbose=1)
    loss,accuracy=model.evaluate(testx,testy,verbose=1)
    predy=model.predict(testx)
    show(history,testx,testy,predy)
def model_b():
    inputs=Input((28,28))
    x1=Flatten()(inputs)
    x2=Dense(128,activation='relu',name='layer1')(x1)
    x3=Dense(256,activation='relu',name='layer2')(x2)
    x3=Dense(128,activation='relu',name='layer3')(x3)
    outputs=Dense(128,activation='softmax',name='layer1')(x1)
    model=Model(inputs,outputs)
    model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
    return model


def show(history,testx,testy,predy):
    plt.figure(figsize=(20,12))
    label=predy.argmax(axis=1)
    for i in range(25):
        plt.subplot(5,5,i+1)
        plt.imshow(testx[i])
        plt.title(f"the original {testy[i]} and the predicted  {label[i]}")
    plt.tight_layout()
    plt.legend()
    plt.show()    
    print("ok")
```

```python
if __name__=='__main__':
    main()
```
