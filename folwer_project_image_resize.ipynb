```python
import cv2
import os

# === CONFIGURATION ===
input_folder = "Project/Image"         # folder where your images are
output_folder = "Project/ResizedImages/" # folder for output
new_size = (256, 256)             # width, height

# === CREATE OUTPUT FOLDER IF NOT EXIST ===
os.makedirs(output_folder, exist_ok=True)

# === LOOP THROUGH ALL FILES ===
for filename in os.listdir(input_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tif")):
        img_path = os.path.join(input_folder, filename)
        image = cv2.imread(img_path)

        if image is None:
            print(f"Skipping {filename} (not readable)")
            continue

        # === RESIZE ===
        resized = cv2.resize(image, new_size)

        # === SAVE TO OUTPUT FOLDER ===
        save_path = os.path.join(output_folder, filename)
        cv2.imwrite(save_path, resized)

        print(f"✅ Resized and saved: {filename}")

print("🎉 All images resized successfully!")
```
