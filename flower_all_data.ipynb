```python
import os
import shutil
import random

# === CONFIGURATION ===
source_folder = "Project/Image/ALL"     # all images in one folder
train_folder = "Project/Image/train"
test_folder = "Project/Image/test"
split_ratio = 0.8  # 80% train, 20% test

# === STEP 1: GROUP IMAGES BY LABEL ===
label_dict = {}

for filename in os.listdir(source_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        label = filename.split("_")[0]  # extract 'IxoraChinensis' from filename
        label_dict.setdefault(label, []).append(filename)

# === STEP 2: SPLIT EACH LABEL INTO TRAIN/TEST ===
for label, images in label_dict.items():
    random.shuffle(images)
    split_index = int(len(images) * split_ratio)
    train_images = images[:split_index]
    test_images = images[split_index:]

    os.makedirs(os.path.join(train_folder, label), exist_ok=True)
    os.makedirs(os.path.join(test_folder, label), exist_ok=True)

    for img in train_images:
        shutil.copy(os.path.join(source_folder, img), os.path.join(train_folder, label, img))
    for img in test_images:
        shutil.copy(os.path.join(source_folder, img), os.path.join(test_folder, label, img))

    print(f"✅ Split {label}: {len(train_images)} train, {len(test_images)} test")

print("🎉 All done: dataset sorted and split!")
```
