```python
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.models import Model
```

```python
def main():
    inputs=Input((8,))
    x1=Dense(4,activation='relu',name='hidden layer1')(inputs)
    x2=Dense(8,activation='relu',name='hidden layer2')(x1)
    x3=Dense(4,activation='relu',name='hidden layer3')(x2)
    outputs=Dense(10,activation='sigmoid',name='output layer')(x3)
    model=Model(inputs,outputs)
    model.summary()
```

```python
if __name__ == '__main__':
    main()
```
