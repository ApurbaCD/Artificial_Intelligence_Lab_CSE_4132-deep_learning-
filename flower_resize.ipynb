```python
import os
from PIL import Image

# === Configuration (edit these) ===
folder_path = "Project/Image/lxora_Chinensis"  # path to your image folder
# class_name = "IxoraCoccinea"                # class name
class_name="IxoraChinensis"                # class name
device = "realme5i"                     # device name
pixel = 12                          # just the number (e.g., 12 for 12MP)
# type_no = 1 
type_no = 2                        # integer (1, 2, etc.)
target_size = (32, 32)               # resize dimensions (width, height)
# ===================================

# Get all image files (ignore directories)
files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

for i, file_name in enumerate(sorted(files), start=1):
    file_ext = os.path.splitext(file_name)[1].lower()
    if file_ext not in ['.jpg', '.jpeg', '.png']:
        continue  # skip non-image files

    # Build new name
    new_name = f"{class_name}_{device}_{pixel}_{type_no}_{i}{file_ext}"
    
    # Paths
    src = os.path.join(folder_path, file_name)
    dst = os.path.join(folder_path, new_name)
    
    # Open and resize image
    img = Image.open(src)
    img = img.resize(target_size)
    img.save(dst)

    # Optionally delete original if name changed
    if src != dst:
        os.remove(src)

print(f"✅ Resized and renamed {len(files)} images in '{folder_path}' to 32×32.")
```
