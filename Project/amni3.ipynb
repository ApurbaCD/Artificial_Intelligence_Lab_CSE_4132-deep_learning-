```python
import os
import shutil
import random

# === CONFIG ===
source_dir = "Project/Image/ALL_P"
target_base = "Project/Image"
class_names = ["IxoraChinensis", "IxoraCoccinea"]
split_ratio = [0.8, 0.1, 0.1]  # train, val, test

# === CREATE TARGET FOLDERS ===
for split in ['train', 'val', 'test']:
    for cls in class_names:
        split_path = os.path.join(target_base, split, cls)
        os.makedirs(split_path, exist_ok=True)

# === SPLIT AND MOVE FILES FOR EACH CLASS ===
for cls in class_names:
    # Filter images belonging to this class
    class_images = [f for f in os.listdir(source_dir) if f.startswith(cls) and f.endswith((".jpg", ".jpeg", ".png"))]
    random.shuffle(class_images)

    total = len(class_images)
    train_end = int(split_ratio[0] * total)
    val_end = train_end + int(split_ratio[1] * total)

    train_imgs = class_images[:train_end]
    val_imgs = class_images[train_end:val_end]
    test_imgs = class_images[val_end:]

    print(len(train_imgs), len(val_imgs), len(test_imgs))

    # Move files
    for img in train_imgs:
        shutil.copy(os.path.join(source_dir, img), os.path.join(target_base, 'train', cls, img))

    for img in val_imgs:
        shutil.copy(os.path.join(source_dir, img), os.path.join(target_base, 'val', cls, img))

    for img in test_imgs:
        shutil.copy(os.path.join(source_dir, img), os.path.join(target_base, 'test', cls, img))

print("✅ Dataset split complete.")
```
