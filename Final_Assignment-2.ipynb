```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense,Input
```

```python
input=Input((8,))
x1=Dense(4,activation='relu',name='layer-1')(input)
x2=Dense(8,activation='relu',name='layer-2')(x1)
x3=Dense(4,activation='relu',name='layer-3')(x2)
output=Dense(1,activation='sigmoid',name='output layer')(x3)
model=Model(input,output)
model.summary(show_trainable='true')
```
