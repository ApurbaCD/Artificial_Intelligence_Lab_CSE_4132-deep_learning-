```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense,Input
```

```python
def main():
    inputs=Input((2,))
    l1=Dense(4,activation='relu',name='Layer1')(inputs)
    l2=Dense(8,activation='relu',name='layer2')(l1)
    l3=Dense(4,activation='relu',name='layer3')(l2)
    outputs=Dense(1,activation='relu',name='output_layer')(l3)
    model=Model(inputs,outputs)
    model.summary(show_trainable=True)
```

```python
if __name__=="__main__":
    main()
```
